```python
# Import the NetworkX package
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
```

## Graph
NetworkX provides several classes to store different types of graphs, such as directed and undirected graph. It also provides classes to create multigraphs (both directed and undirected).

For more information, please refer to [NetworkX graph types](https://networkx.org/documentation/stable/reference/classes/index.html).

```python
# Create an undirected graph G
G = nx.Graph()
print(G.is_directed())

# Create a directed graph H
H = nx.DiGraph()
print(H.is_directed())

# Add graph level attribute
G.graph["Name"] = "Bar"
print(G.graph)
```

## Node

Nodes (with attributes) can be easily added to NetworkX graphs.

```python
# Add one node with node level attributes
G.add_node(0, feature=5, label=0)

# Get attributes of the node 0
node_0_attr = G.nodes[0]
print("Node 0 has the attributes {}".format(node_0_attr))
```

```python
G.nodes(data=True)
```

```python
# Add multiple nodes with attributes
G.add_nodes_from([
  (1, {"feature": 1, "label": 1}),
  (2, {"feature": 2, "label": 2})
]) #(node, attrdict)

# Loop through all the nodes
# Set data=True will return node attributes
for node in G.nodes(data=True):
  print(node)

# Get number of nodes
num_nodes = G.number_of_nodes()
print("G has {} nodes".format(num_nodes))
```

## Edge

Similar to nodes, edges (with attributes) can also be easily added to NetworkX graphs.

```python
# Add one edge with edge weight 0.5
G.add_edge(0, 1, weight=0.5)

# Get attributes of the edge (0, 1)
edge_0_1_attr = G.edges[(0, 1)]
print("Edge (0, 1) has the attributes {}".format(edge_0_1_attr))
```

```python
# Add multiple edges with edge weights
G.add_edges_from([
  (1, 2, {"weight": 0.3}),
  (2, 0, {"weight": 0.1})
])

# Loop through all the edges
# Here there is no data=True, so only the edge will be returned
for edge in G.edges():
  print(edge)

# Get number of edges
num_edges = G.number_of_edges()
print("G has {} edges".format(num_edges))
```

## Visualization

```python
# Draw the graph
nx.draw(G, with_labels = True)
```

## Node Degree and Neighbor

```python
node_id = 1

# Degree of node 1
print("Node {} has degree {}".format(node_id, G.degree[node_id]))

# Get neighbor of node 1
for neighbor in G.neighbors(node_id):
  print("Node {} has neighbor {}".format(node_id, neighbor))
```

## Random Graph creation and analysis

```python
# creation of a random network with N nodes and E edges - this a erdos-renyi graph
def create(n,m):
    G = nx.gnm_random_graph(n, m, seed = random_state)
    return G
```

```python
def degree_analysis(G):
    degree_sequence = sorted((d for n, d in G.degree()), reverse=True)

    fig = plt.figure("Degree of a random graph", figsize=(8, 8))
    # Create a gridspec for adding subplots of different sizes
    axgrid = fig.add_gridspec(2, 4)


    ax1 = fig.add_subplot(axgrid[:, :2])
    ax1.plot(degree_sequence, "b-", marker="o")
    ax1.set_title("Degree Rank Plot")
    ax1.set_ylabel("Degree")
    ax1.set_xlabel("Rank")

    ax2 = fig.add_subplot(axgrid[:, 2:])
    ax2.bar(*np.unique(degree_sequence, return_counts=True))
    ax2.set_title("Degree histogram")
    ax2.set_xlabel("Degree")
    ax2.set_ylabel("# of Nodes")

    plt.show()
```

```python
random_state = np.random.RandomState(42)
N=20 #number of nodes
E=2*N #number of edges
G=create(N,E)
```

```python
print(G.edges(nbunch=4))
print(G.adj)
print(G.edges)
print(type(G))
print(len(G.edges))
```

```python
pos_cons = nx.spring_layout(G,seed = random_state)
nx.draw(G, pos=pos_cons,labels ={node:node for node in G.nodes()}) #spring_layout , spectral_layout
plt.draw()
```

```python
# Get the adjacency matrix
A = nx.adjacency_matrix(G).toarray()
print("Adjacency Matrix:")
print(A)



# Get the degree matrix
degree_sequence_full = [d for n, d in G.degree()]
D = np.diag(degree_sequence_full)
print("\nDegree Matrix:")
print(D)



# Get the Laplacian matrix
L = D - A
print("\nLaplacian Matrix:")
print(L)
```

```python
degree_analysis(G)
```

```python
# calculate the graph density for the above created random graph
V = G.number_of_nodes()
E_count = G.number_of_edges()
density = (2 * E_count) / (V * (V - 1))
print(f"Number of nodes |V| = {V}")
print(f"Number of edges |E| = {E_count}")
print(f"Graph density D = {density:.5f}")
```

```python
# increase N from 20 to 200 and see how density and degree distribution change
N_values = [20, 40, 60, 80, 100, 120, 150, 180, 200]
densities = []
mean_degrees = []

fig, axes = plt.subplots(3, 3, figsize=(16, 14))
axes = axes.flatten()

for i, n in enumerate(N_values):
    random_state = np.random.RandomState(42)
    m = 2 * n  # keep E = 2N like the original graph
    Gn = create(n, m)

    d = (2 * Gn.number_of_edges()) / (n * (n - 1))
    densities.append(d)

    deg_seq = [deg for node, deg in Gn.degree()]
    mean_degrees.append(np.mean(deg_seq))

    axes[i].bar(*np.unique(deg_seq, return_counts=True))
    axes[i].set_title(f"N={n}, density={d:.4f}")
    axes[i].set_xlabel("Degree")
    axes[i].set_ylabel("# of Nodes")

plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
plt.plot(N_values, densities, marker='o')
plt.xlabel("N (number of nodes)")
plt.ylabel("Graph density D")
plt.title("Graph density vs N")
plt.show()

for n, d, md_ in zip(N_values, densities, mean_degrees):
    print(f"N={n:4d}  density={d:.5f}  mean_degree={md_:.2f}")
```

As N increases from 20 to 200, the density drops a lot (from about 0.21 down to 0.02), even though the average degree stays about the same (~4) since we kept E = 2N. This makes sense because density depends on N^2 in the denominator, so it shrinks fast as the graph gets bigger, even if each node still has roughly the same number of neighbours.

The degree histogram also looks more like a smooth bell curve at N=200 compared to N=20, where it's more random/uneven just because there are fewer nodes to average over.
